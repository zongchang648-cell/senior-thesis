```python
import requests
import pandas as pd
import time
from urllib.parse import quote

# ===== 配置 =====
input_file = r"C:\毕业论文\数据\drug_list.txt"
output_file = r"C:\毕业论文\数据\drug_smiles.csv"
# API 调用间隔（秒），避免触发频率限制（建议 0.2-0.5 秒）
delay = 0.3
# ================

def get_smiles_from_name(drug_name):
    """
    根据药物名称从 PubChem 获取 SMILES
    返回 SMILES 字符串，如果失败则返回 None
    """
    # URL 编码药物名称（处理空格、括号等）
    encoded_name = quote(drug_name.strip())
    url = f"https://pubchem.ncbi.nlm.nih.gov/rest/pug/compound/name/{encoded_name}/property/IsomericSMILES/CSV"
    
    try:
        response = requests.get(url, timeout=10)
        if response.status_code == 200:
            # 返回的 CSV 第一行是标题，第二行是数据
            lines = response.text.strip().split('\n')
            if len(lines) >= 2:
                # 第二行格式：CID,IsomericSMILES,...
                smiles = lines[1].split(',')[1]  # 第二列是 SMILES
                return smiles
        elif response.status_code == 404:
            # 未找到化合物
            return None
        else:
            print(f"警告：{drug_name} 返回状态码 {response.status_code}")
            return None
    except Exception as e:
        print(f"请求 {drug_name} 时出错：{e}")
        return None

def main():
    # 读取药物列表
    with open(input_file, 'r', encoding='utf-8') as f:
        drug_names = [line.strip() for line in f if line.strip()]
    
    print(f"共读取到 {len(drug_names)} 个药物名称")
    
    results = []
    for i, name in enumerate(drug_names, 1):
        print(f"正在处理 ({i}/{len(drug_names)})：{name}")
        smiles = get_smiles_from_name(name)
        results.append({
            "Drug Name": name,
            "SMILES": smiles if smiles else "Not Found"
        })
        time.sleep(delay)  # 暂停，避免请求过快
    
    # 保存为 CSV
    df = pd.DataFrame(results)
    df.to_csv(output_file, index=False, encoding='utf-8-sig')
    print(f"完成！结果已保存至：{output_file}")

if __name__ == "__main__":
    main()
```
